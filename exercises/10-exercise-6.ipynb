> Load the FLORES-200 devtest sentences for the eleven NLLB Turkic varieties (see the chapter’s NLLB language-code table), or the X-WMT test sets for the six languages that X-WMT and NLLB share. Align by documented sentence IDs or verify correspondence from the release; do not assume that independently filtered files remain row-aligned. For an exploratory subset, select and record the sentence IDs (for example 200) and translate between distinct-language pairs with `facebook/nllb-200-distilled-600M` (one TurkicNLP `translate` pipeline per source language, with `translate_tgt_lang` set to each target) and compute chrF against the references. Plot the matrix as a heatmap. Label the matrix as a subset experiment, exclude identity directions, and tune choices only on development data. Do within-branch pairs score higher than cross-branch pairs? Do same-script pairs score higher than cross-script pairs with similar genealogical distance?

**Setup.** `pip install "turkicnlp[hfst]==0.5.0" torch transformers sentencepiece sacrebleu requests` (the compiled `hfst` package needs Python 3.11 or 3.12; elsewhere install `pyhfst`). NLLB-200 distilled 600M is loaded from `NLLB_PATH` (a local copy) or from the Hugging Face Hub at the revision pinned in the book; it runs in float32 on the CPU, about 1.7 s per sentence on four cores. Translations are cached in `BOOK_DATA_DIR` (default `~/.cache/turkic-nlp-book`), so an interrupted run resumes where it stopped. Test data come from the commit-pinned TIL release (X-WMT test sets, MIT licence); the X-WMT files are tokenised, and the code removes the spaces around punctuation before scoring (a heuristic, see `detok`). Scores use SacreBLEU with its default settings: BLEU with the `13a` tokeniser, chrF, and chrF++ (`word_order=2`).

**Subset.** The original FLORES-200 archive is distributed from a host that the environment of the worked solution could not reach, so the matrix uses X-WMT: Azerbaijani, Turkish, Uzbek, Kazakh, Kyrgyz and Bashkir, the six languages in both X-WMT and NLLB-200. Each of the 30 directions uses the first 15 sentences of its own X-WMT file (set `MATRIX_N` for more). The X-WMT directions do not all contain the same sentences, so cells are not computed on identical content, unlike a FLORES matrix. No setting was tuned.

````python
import collections, hashlib, os, re, statistics, time
from pathlib import Path
import requests

# Commit-pinned TIL release files (X-WMT test sets and the corpus README); MIT licence.
TIL = "https://raw.githubusercontent.com/turkic-interlingua/til-mt/7d988cc47dba889d02262d893bf9a46e43a6bb36/"
CACHE = Path(os.environ.get("BOOK_DATA_DIR", "~/.cache/turkic-nlp-book")).expanduser()
FETCHED = {}                                   # url -> sha256 of every file read (saved with the outputs)

def fetch(url):
    path = CACHE / hashlib.sha1(url.encode()).hexdigest()[:12] / url.rsplit("/", 1)[-1]
    if not path.exists():
        r = requests.get(url, timeout=300)
        r.raise_for_status()
        path.parent.mkdir(parents=True, exist_ok=True)
        path.write_bytes(r.content)
    FETCHED[url] = hashlib.sha256(path.read_bytes()).hexdigest()
    return path

def detok(s):
    """Undo the X-WMT tokenisation (spaces around punctuation) so references match raw MT output."""
    s = re.sub(r" ([’'] ?)", lambda m: m.group(1).strip(), s)       # Sialkot ’ taki -> Sialkot’taki
    s = re.sub(r" ([.,!?;:%)\]»])", r"\1", s)
    return re.sub(r"([(\[«]) ", r"\1", s).strip()

def xwmt(src, tgt, n=None):
    """The first n sentence pairs of the X-WMT test set for src-tgt (X-WMT language codes)."""
    s = fetch(f"{TIL}xwmt/test/{src}-{tgt}/{src}-{tgt}.{src}").read_text(encoding="utf-8").splitlines()
    t = fetch(f"{TIL}xwmt/test/{src}-{tgt}/{src}-{tgt}.{tgt}").read_text(encoding="utf-8").splitlines()
    assert len(s) == len(t), (src, tgt)
    pairs = [(detok(a), detok(b)) for a, b in zip(s, t) if a.strip() and b.strip()]
    return pairs[:n] if n else pairs

NLLB_CODE = {"az": "azj_Latn", "ba": "bak_Cyrl", "kk": "kaz_Cyrl", "ky": "kir_Cyrl", "tr": "tur_Latn",
             "uz": "uzn_Latn", "en": "eng_Latn", "ru": "rus_Cyrl"}
NLLB = {}

def nllb():
    """NLLB-200 distilled 600M: from NLLB_PATH (a local copy) or the pinned Hub revision; float32, CPU."""
    if not NLLB:
        import torch
        from transformers import AutoModelForSeq2SeqLM, NllbTokenizer
        path = os.environ.get("NLLB_PATH", "facebook/nllb-200-distilled-600M")
        rev = None if Path(path).exists() else "f8d333a098d19b4fd9a8b18f94170487ad3f821d"
        NLLB["tok"] = NllbTokenizer.from_pretrained(path, revision=rev)
        NLLB["model"] = AutoModelForSeq2SeqLM.from_pretrained(path, revision=rev, torch_dtype=torch.float32).eval()
        NLLB["torch"] = torch
        torch.set_num_threads(os.cpu_count() or 1)
    return NLLB["tok"], NLLB["model"]

CACHE_FILE = CACHE / "nllb-translations.jsonl"       # translations are cached, so a run can resume
LAST_SECONDS = 0.0

def _cached():
    import json
    out = {}
    if CACHE_FILE.exists():
        for line in CACHE_FILE.read_text(encoding="utf-8").splitlines():
            r = json.loads(line); out[tuple(r["key"])] = (r["out"], r["capped"], r["sec"])
    return out

def translate(sents, src, tgt, beams=1, batch=8, max_new=256):
    """Translate with explicit source and target language tokens (greedy unless beams > 1).
    Returns (outputs, number of outputs that reached the generation ceiling without an
    end-of-sequence token). Results are cached per sentence in CACHE_FILE, with the generation
    time of their batch divided by its size; LAST_SECONDS holds the total for the last call."""
    global LAST_SECONDS
    import json
    cache = _cached()
    key = lambda s: [src, tgt, beams, max_new, s]
    todo = [s for s in dict.fromkeys(sents) if tuple(key(s)) not in cache]
    if todo:
        tok, model = nllb()
        tok.src_lang = NLLB_CODE[src]
        CACHE_FILE.parent.mkdir(parents=True, exist_ok=True)
        with NLLB["torch"].inference_mode(), CACHE_FILE.open("a", encoding="utf-8") as f:
            for i in range(0, len(todo), batch):
                chunk = todo[i:i + batch]
                t0 = time.time()
                enc = tok(chunk, return_tensors="pt", padding=True)
                gen = model.generate(**enc, forced_bos_token_id=tok.convert_tokens_to_ids(NLLB_CODE[tgt]),
                                     num_beams=beams, do_sample=False, max_new_tokens=max_new)
                sec = (time.time() - t0) / len(chunk)
                for s, row, text in zip(chunk, gen, tok.batch_decode(gen, skip_special_tokens=True)):
                    capped = int(row[-1] != tok.eos_token_id and row[-1] != tok.pad_token_id)
                    cache[tuple(key(s))] = (text, capped, sec)
                    f.write(json.dumps({"key": key(s), "out": text, "capped": capped, "sec": sec}, ensure_ascii=False) + "\n")
                f.flush()
    res = [cache[tuple(key(s))] for s in sents]
    LAST_SECONDS = sum(r[2] for r in res)
    return [r[0] for r in res], sum(r[1] for r in res)

def metrics(hyps, refs):
    """Corpus BLEU (13a), chrF and chrF++ with SacreBLEU, rounded, plus the chrF++ signature."""
    from sacrebleu.metrics import BLEU, CHRF
    bleu, chrf, chrfpp = BLEU(tokenize="13a"), CHRF(), CHRF(word_order=2)
    r = {"BLEU": bleu.corpus_score(hyps, [refs]).score, "chrF": chrf.corpus_score(hyps, [refs]).score,
         "chrF++": chrfpp.corpus_score(hyps, [refs]).score}
    return {k: round(v, 2) for k, v in r.items()}, str(chrfpp.get_signature())
````

````python
LANGS = ["az", "tr", "uz", "kk", "ky", "ba"]          # in both X-WMT and NLLB-200
BRANCH = {"az": "Oghuz", "tr": "Oghuz", "uz": "Karluk", "kk": "Kipchak", "ky": "Kipchak", "ba": "Kipchak"}
SCRIPT = {"az": "Latn", "tr": "Latn", "uz": "Latn", "kk": "Cyrl", "ky": "Cyrl", "ba": "Cyrl"}
N = int(os.environ.get("MATRIX_N", 20))
matrix, seconds = {}, 0.0
for s in LANGS:
    for t in LANGS:
        if s == t:
            continue
        pairs = xwmt(s, t, N)
        hyp, _ = translate([a for a, _ in pairs], s, t); seconds += LAST_SECONDS
        matrix[s, t] = metrics(hyp, [b for _, b in pairs])[0]["chrF"]
print(f"chrF on the first {N} X-WMT sentences of each direction ({seconds:.0f} s generation time); rows = source")
print("      " + "".join(f"{t:>7}" for t in LANGS))
for s in LANGS:
    print(f"{s:>4}  " + "".join(f"{matrix[s, t]:7.1f}" if s != t else "      -" for t in LANGS))
groups = collections.defaultdict(list)
for (s, t), v in matrix.items():
    groups[("same branch" if BRANCH[s] == BRANCH[t] else "cross branch") + ", " +
           ("same script" if SCRIPT[s] == SCRIPT[t] else "cross script")].append(v)
for k, v in sorted(groups.items()):
    print(f"  {k:30} {len(v):2} directions, mean chrF {statistics.mean(v):.1f}")
````

## Worked solution

Output recorded in `experiments/ch10_exercises/verified-output.json` (NLLB-200 distilled 600M, CPU, float32; **subset experiment: 15 sentences per direction**):

```
chrF on the first 15 X-WMT sentences of each direction (798 s generation time); rows = source
           az     tr     uz     kk     ky     ba
  az        -   46.1   44.5   45.1   44.9   34.7
  tr     48.8      -   38.5   38.1   40.1   31.3
  uz     49.4   41.6      -   51.5   43.0   35.4
  kk     41.8   38.1   47.7      -   41.1   33.6
  ky     37.6   37.9   39.6   42.0      -   45.2
  ba     40.2   36.1   42.8   41.1   56.6      -
  cross branch, cross script     18 directions, mean chrF 40.3
  cross branch, same script       4 directions, mean chrF 43.5
  same branch, same script        8 directions, mean chrF 44.3
```

**The pattern.** Same-branch directions average chrF 44.3, cross-branch directions 40.3 (cross-script) and 43.5 (same script). The highest cell is Bashkir→Kyrgyz (56.6), within Kipchak; the next is Uzbek→Kazakh (51.5), across branches. Translation *into* Bashkir is the weakest column (31.3–45.2).

**What the matrix can and cannot show.** In this language set branch and script are almost perfectly confounded: the Oghuz languages are written in Latin, the Kipchak ones in Cyrillic, and only Uzbek (Karluk, Latin) separates the two factors. The "same branch, cross script" group is therefore empty, and the comparison of same-script with cross-script pairs at a similar genealogical distance rests on the four Uzbek–Oghuz directions alone. With 15 sentences per cell, different sentences in different cells and chrF differences of a few points, the matrix is a hypothesis generator, not evidence that relatedness or script drives quality. A larger matrix should use aligned sentence IDs and language/script conditions actually supported by the model and references. Crimean Tatar and Uyghur add contrasts in NLLB; converted Azerbaijani or Kazakh input would be a separate intervention, not an additional native-script NLLB condition. Neither expansion alone controls training-data and domain differences.
