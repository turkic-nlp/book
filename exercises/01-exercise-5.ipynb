# Exercise 5: What a toolkit offers per language

*The Turkic World · version 0.1.0*

> Install TurkicNLP and run `turkicnlp.list_processors(lang)` for `tur`, `kaz`, `kir`, `sah`, `kum` and `chv`. Which processors are available for each language, and from which backends? Compare the result with the resource table: which tasks have both a listed backend and suitable evaluation data?

**Setup.** `pip install "turkicnlp==0.5.0" pandas`. No downloads are needed to list processors.

```python
import pandas as pd
import turkicnlp

LANGS = ["tur", "kaz", "kir", "sah", "kum", "chv"]
print("TurkicNLP", turkicnlp.__version__)
rows = []
for lang in LANGS:
    for processor, backends in turkicnlp.list_processors(lang).items():
        rows.append({"language": lang, "processor": processor, "backends": ", ".join(backends)})
table = pd.DataFrame(rows)
table.pivot(index="processor", columns="language", values="backends").fillna("—")
```

```python
# Evaluation data for the main tasks, from the book (Neural Morphology, Syntax and Universal Dependencies, and Sequence Labelling and the KazNERD/UD inventories).
# "—" means no evaluation set is named in the book for that language; it does not prove that none exists.
EVAL_DATA = {
    ("tur", "depparse"): "UD Turkish treebanks (e.g. IMST, BOUN)",
    ("tur", "pos"): "UD Turkish treebanks (e.g. IMST, BOUN)",
    ("kaz", "pos"): "UD Kazakh-KTB (note: split-exposure caveat, Syntax and Universal Dependencies)",
    ("kir", "pos"): "UD Kyrgyz-KTMU and TueCL",
    ("sah", "pos"): "small UD-Turkic parallel test set (Syntax and Universal Dependencies)",
    ("sah", "depparse"): "small UD-Turkic parallel test set (Syntax and Universal Dependencies)",
    ("kum", "pos"): "small UD-Turkic parallel test set (Syntax and Universal Dependencies)",
    ("kum", "depparse"): "small UD-Turkic parallel test set (Syntax and Universal Dependencies)",
    ("kaz", "depparse"): "UD Kazakh-KTB (note: split-exposure caveat, Syntax and Universal Dependencies)",
    ("kir", "depparse"): "UD Kyrgyz-KTMU and TueCL",
    ("tur", "ner"): "Turkish NER datasets (e.g. Mukayese)",
    ("kaz", "ner"): "KazNERD",
    ("tur", "translate"): "FLORES-200, X-WMT",
    ("kaz", "translate"): "FLORES-200, X-WMT",
    ("kir", "translate"): "FLORES-200, X-WMT",
}
check = table[table["processor"].isin(["pos", "depparse", "ner", "translate", "morph"])].copy()
check["evaluation data named in the book"] = [EVAL_DATA.get((l, p), "—") for l, p in zip(check["language"], check["processor"])]
check.sort_values(["processor", "language"]).reset_index(drop=True)
```

## Worked solution (TurkicNLP 0.5.0)

- **Turkish and Kazakh** have the fullest coverage: Apertium morphology, Stanza tokenisation, tagging, lemmatisation, parsing and NER, the multilingual Glot500 tagger/parser/morphology, NLLB translation and embeddings, and Omnilingual speech recognition.
- **Kyrgyz** has the same, except NER.
- **Sakha and Kumyk** have Apertium morphology, a rule-based tokeniser, the multilingual Glot500 tagger, parser and morphology, and speech recognition, but no Stanza models, NER or NLLB translation.
- **Chuvash** has only morphology (Apertium and Glot500), a rule-based tokeniser and speech recognition: no tagger or parser.

**Backend and evaluation data together.** Parsing for Turkish, Kazakh and Kyrgyz (UD treebanks), NER for Turkish and Kazakh (KazNERD for Kazakh), and translation for the NLLB languages (FLORES-200, X-WMT) have both a backend and a named evaluation set. For Sakha and Kumyk, the multilingual tagger and parser run, but the book's only gold data are small parallel test sets (Syntax and Universal Dependencies), so their quality is largely unmeasured. For Chuvash, a listed morphology backend is not yet evidence of accuracy: an independent held-out evaluation is needed (Morphological Structure). A listed processor is a starting point to evaluate, not a measured capability.
